# Pandas

Pandas is a powerful python library for working with structured data, enabling tasks like importing, cleansing, transforming and analyzing datasets efficiently and make conclusions based on statistical theories.

- **Series**: A Series is a one-dimensional labeled array that can hold any data type (integer, string, float, Python objects, etc.).
- **Data Frames**: A Data Frame is a 2-dimensional tabular data structure with labeled axes (rows and columns). A DataFrame object is similar to a spreadsheet or a database table.
- **Panel (Deprecated)**: A Panel is a 3-dimensional data structure, similar to a DataFrame, but with three dimensions (rows, columns, and pages).
- **Index**: An Index object is a one-dimensional array that can hold any type of data (integer, string, float, Python objects, etc.).
- **MultiIndex**: A MultiIndex object is a hierarchical index, similar to an Index, but with multiple levels.
- **Categorical**: A Categorical object is a type of Index that represents categorical data.
- **DatetimeIndex**: A DatetimeIndex object is a type of Index that represents datetime data.
- **TimedeltaIndex**: A TimedeltaIndex object is a type of Index that represents timedelta data.
- **IntervalIndex**: An IntervalIndex object is a type of Index that represents interval data.
- **Sparse**: A Sparse object is a sparse representation of a data structure, such as a DataFrame or Series.
- **Panel4D**: A Panel4D object is a 4-dimensional data structure, similar to a Panel, but with four dimensions (rows, columns, pages, and sections).

```python
import pandas as pd
```

## Series

```python
# Create a Series
s = pd.Series([1, 2, 3, 4, 5])
print(s.values)  # Print the values of the Series.
print('Mean:', s.mean())  # Calculate the mean of the Series.
print('Median:', s.median())  # Calculate the median of the Series.
print('Mode:', s.mode())  # Calculate the mode of the Series.
print('Standard Deviation:', s.std())  # Calculate the standard deviation of the Series.
print('Variance:', s.var())  # Calculate the variance of the Series.
```

## DataFrame

```python
# DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 40, 45],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}
df = pd.DataFrame(data)
print(df)  # Print the DataFrame.
print(df['Name'])  # Access the 'Name' column.
print(df['Age'].mean())  # Calculate the mean age.
print(df['City'].unique())  # Get unique cities.
print(df.describe())  # Get summary statistics for the DataFrame.
print(df[df['Age'] > 30])  # Filter rows where age is greater than 30.
print(df.sort_values(by='Age'))  # Sort the DataFrame by age.
print(df.groupby('City')['Age'].mean())  # Group by city and calculate mean age.
print(df.isnull().sum())  # Check for missing values in the DataFrame.
print(df['Age'].fillna(df['Age'].mean()))  # Fill missing values in 'Age' with the mean age.
print(df['City'].value_counts())  # Count the occurrences of each city.
print(df['Age'].apply(lambda x: 'Adult' if x >= 18 else 'Minor'))  # Apply a function to the 'Age' column.
print(df['Name'].str.upper())  # Convert the 'Name' column to uppercase.
print(df['City'].str.contains('New'))  # Check if 'City' contains 'New'.
# Ensure correlation is computed only on numeric columns (avoid converting string columns like 'Name' to float)
print(df.corr(numeric_only=True))  # Calculate the correlation between numeric columns in the DataFrame.
```

## Index

```python
# Create a Index
index = pd.Index(['a', 'b', 'c', 'd', 'e'])
print(index)  # Print the Index.
print(index[0])  # Access the first element of the Index.
print(index[1:4])  # Access a slice of the Index.
print(index.append(pd.Index(['f', 'g'])))  # Append another Index to the current Index.
print(index.intersection(pd.Index(['b', 'c', 'f'])))  # Find the intersection of the current Index with another Index.
print(index.difference(pd.Index(['b', 'c'])))  # Find the difference between the current Index and another Index.
print(index.union(pd.Index(['f', 'g'])))  # Find the union of the current Index with another Index.
print(index.isin(['a', 'e', 'f']))  # Check if elements of the Index are in a list of values.
print(index.delete(2))  # Delete the element at index position 2.
print(index.insert(2, 'x'))  # Insert a new element 'x' at index position 2.
print(index.rename('MyIndex'))  # Rename the Index.
print(index.to_list())  # Convert the Index to a list.
print(index.to_series())  # Convert the Index to a Series.
print(index.to_frame())  # Convert the Index to a DataFrame.
print(index.value_counts())  # Count the occurrences of each unique value in the Index.
print(index.nunique())  # Count the number of unique values in the Index.
print(index.is_unique)  # Check if all values in the Index are unique.
print(index.is_monotonic_increasing or index.is_monotonic_decreasing)  # Check if the Index is monotonic (sorted).
print(index.is_monotonic_increasing)  # Check if the Index is monotonically increasing.
print(index.is_monotonic_decreasing)  # Check if the Index is monotonically decreasing
print(index.has_duplicates)  # Check if the Index has duplicate values.
print(index.isin(['a', 'b', 'c']))  # Check if elements of the Index are in a list of values.
print(index.get_loc('c'))  # Get the location of the element 'c' in the Index.
print(index.get_indexer(['a', 'c', 'e']))  # Get the indexer for a list of values in the Index.
print(index.get_indexer_for(['a', 'c', 'e']))  # Get the indexer for a list of values in the Index.
print(index.get_indexer_non_unique(['a', 'c', 'e']))  # Get the indexer for a list of values in the Index, including non-unique values.
```

# Categorical

```python
# Categorical data
categories = pd.Categorical(['apple', 'banana', 'apple', 'orange', 'banana'])
print(categories)  # Print the Categorical data.
print(categories.categories)  # Print the categories.
print(categories.codes)  # Print the codes for the Categorical data.
print(categories.value_counts())  # Count the occurrences of each category.
print(categories.describe())  # Get summary statistics for the Categorical data.
print(categories.isnull())  # Check for missing values in the Categorical data.
print(categories.notnull())  # Check for non-missing values in the Categorical data.
print(categories.astype('object'))  # Convert the Categorical data to an object type.
print(categories.astype('str'))  # Convert the Categorical data to a string type.
print(categories.astype('category'))  # Convert the Categorical data to a category type (no change).
print(categories.add_categories(['grape']))  # Add a new category 'grape' to the Categorical data.
print(categories.remove_categories(['banana']))  # Remove the category 'banana' from the Categorical data.
print(categories.rename_categories({'apple': 'fruit', 'orange': 'citrus'}))  # Rename categories in the Categorical data.
print(categories.set_categories(['fruit', 'citrus', 'grape'], ordered=True))  # Set new categories and order for the Categorical data.
print(categories.as_ordered())  # Set the Categorical data as ordered.
print(categories.as_unordered())  # Set the Categorical data as unordered.
print(categories.ordered)  # Check if the Categorical data is ordered.
print(categories.unique)  # Check the unique values in the Categorical data.
```

## DatetimeIndex

```python
# DatetimeIndex
dates = pd.date_range('2024-01-01', periods=5, freq='D')
print(dates)  # Print the DatetimeIndex.
print(dates.year)  # Access the year component of the DatetimeIndex.
print(dates.month)  # Access the month component of the DatetimeIndex.
print(dates.day)  # Access the day component of the DatetimeIndex.
print(dates.weekday)  # Access the weekday component of the DatetimeIndex (0=Monday, 6=Sunday).
print(dates.dayofweek)  # Access the day of the week component of the DatetimeIndex (0=Monday, 6=Sunday).
print(dates.dayofyear)  # Access the day of the year component of the DatetimeIndex (1-366).
print(dates.is_month_start)  # Check if each date is the start of the month.
print(dates.is_month_end)  # Check if each date is the end of the month.
print(dates.is_quarter_start)  # Check if each date is the start of the quarter.
print(dates.is_quarter_end)  # Check if each date is the end of the quarter.
print(dates.is_year_start)  # Check if each date is the start of the year.
print(dates.is_year_end)  # Check if each date is the end of the year.
print(dates.to_period('M'))  # Convert the DatetimeIndex to a PeriodIndex with monthly frequency.
print(dates.to_period('Q'))  # Convert the DatetimeIndex to a PeriodIndex with quarterly frequency.
print(dates.to_period('Y'))  # Convert the DatetimeIndex to a PeriodIndex with annual frequency.
print(dates.to_julian_date())  # Convert the DatetimeIndex to Julian dates.
print(dates.to_pydatetime())  # Convert the DatetimeIndex to an array of Python datetime objects.
print(dates.to_numpy())  # Convert the DatetimeIndex to a NumPy array of datetime64 objects.
print(dates.to_series())  # Convert the DatetimeIndex to a Series of Timestamps.
print(dates.to_frame())  # Convert the DatetimeIndex to a DataFrame with a single column of Timestamps.
print(dates.isin(pd.date_range('2024-01-02', '2024-01-04')))  # Check if dates are in a specified date range.
print(dates.get_loc('2024-01-03'))  # Get the location of a specific date in the DatetimeIndex.
print(dates.get_indexer(pd.date_range('2024-01-02', '2024-01-04')))  # Get the indexer for a specified date range in the DatetimeIndex.
print(dates.get_indexer_for(pd.date_range('2024-01-02', '2024-01-04')))  # Get the indexer for a specified date range in the DatetimeIndex, including non-unique values.
print(dates.get_indexer_non_unique(pd.date_range('2024-01-02', '2024-01-04')))  # Get the indexer for a specified date
```

## TimedeltaIndex

```python
## TimedeltaIndex
timedeltas = pd.to_timedelta(['1 day', '2 days', '3 days', '4 days', '5 days'])
print(timedeltas)  # Print the TimedeltaIndex.
print(timedeltas.days)  # Access the days component of the TimedeltaIndex.
print(timedeltas.total_seconds())  # Access the total seconds component of the TimedeltaIndex.
print(timedeltas.components)  # Access the components (days, hours, minutes, seconds) of the TimedeltaIndex.
print(timedeltas.to_pytimedelta())  # Convert the TimedeltaIndex to an array of Python timedelta objects.
print(timedeltas.to_numpy())  # Convert the TimedeltaIndex to a NumPy array of timedelta64 objects.
print(timedeltas.to_series())  # Convert the TimedeltaIndex to a Series of Timedelta objects.
print(timedeltas.to_frame())  # Convert the TimedeltaIndex to a DataFrame with a single column of Timedelta objects.
print(timedeltas.isin(pd.to_timedelta(['2 days', '4 days'])))  # Check if timedeltas are in a specified list of timedeltas.
print(timedeltas.get_loc(pd.to_timedelta('3 days')))  # Get the location of a specific timedelta in the TimedeltaIndex.
print(timedeltas.get_indexer(pd.to_timedelta(['2 days', '4 days'])))  # Get the indexer for a specified list of timedeltas in the TimedeltaIndex.
print(timedeltas.get_indexer_for(pd.to_timedelta(['2 days', '4 days'])))  # Get the indexer for a specified list of timedeltas in the TimedeltaIndex, including non-unique values.
print(timedeltas.get_indexer_non_unique(pd.to_timedelta(['2 days', '4 days'])))  # Get the indexer for a specified list of timedeltas in the TimedeltaIndex, including non-unique values.
```

## IntervalIndex

```python
# IntervalIndex
intervals = pd.IntervalIndex.from_tuples([(0, 1), (1, 2), (2, 3), (3, 4), (4, 5)])
print(intervals)  # Print the IntervalIndex.
print(intervals.left)  # Access the left endpoints of the intervals.
print(intervals.right)  # Access the right endpoints of the intervals.
print(intervals.closed)  # Access the closed attribute of the intervals (whether they are closed on the left, right, both, or neither).
print(intervals.to_series())  # Convert the IntervalIndex to a Series of Interval objects.
print(intervals.to_frame())  # Convert the IntervalIndex to a DataFrame with two columns for the left and right endpoints of the intervals.
print(intervals.isin([pd.Interval(1, 2), pd.Interval(3, 4)]))  # Check if specific intervals are in the IntervalIndex.
print(intervals.get_loc(pd.Interval(2, 3)))  # Get the location of a specific interval in the IntervalIndex.
print(intervals.get_indexer(pd.IntervalIndex([pd.Interval(1, 2), pd.Interval(3, 4)])))  # Get the indexer for a specified list of intervals in the IntervalIndex.
# Note: get_indexer_for and get_indexer_non_unique are not standard methods for IntervalIndex; removing them to avoid errors.
```

```python
# PeriodIndex
periods = pd.period_range('2024-01', periods=5, freq='M')
print(periods)  # Print the PeriodIndex.
print(periods.year)  # Access the year component of the PeriodIndex.
print(periods.month)  # Access the month component of the PeriodIndex.
print(periods.day)  # Access the day component of the PeriodIndex (will be 1 for monthly frequency).
print(periods.quarter)  # Access the quarter component of the PeriodIndex.
print(periods.to_timestamp())  # Convert the PeriodIndex to a DatetimeIndex.
print(periods.to_timestamp(how='end'))  # Convert the PeriodIndex to a DatetimeIndex, using the end of the period.
print(periods.to_timestamp(how='start'))  # Convert the PeriodIndex to a DatetimeIndex, using the start of the period.
print(periods.to_timestamp(how='e'))  # Convert the PeriodIndex to a DatetimeIndex, using the end of the period.
print(periods.to_timestamp(how='s'))  # Convert the PeriodIndex to a DatetimeIndex, using the start of the period.
print(periods.to_series())  # Convert the PeriodIndex to a Series of Period objects.
print(periods.to_frame())  # Convert the PeriodIndex to a DataFrame with a single column of Period objects.
print(periods.isin(pd.period_range('2024-02', '2024-04', freq='M')))  # Check if periods are in a specified range of periods.
print(periods.get_loc(pd.Period('2024-03', freq='M')))  # Get the location of a specific period in the PeriodIndex.
print(periods.get_indexer(pd.period_range('2024-02', '2024-04', freq='M')))  # Get the indexer for a specified range of periods in the PeriodIndex.
print(periods.get_indexer_for(pd.period_range('2024-02', '2024-04', freq='M'))  # Get the indexer for a specified range of periods in the PeriodIndex, including non-unique values.
      )
print(periods.get_indexer_non_unique(pd.period_range('2024-02', '2024-04', freq='M'))  # Get the indexer for a specified range of periods in the PeriodIndex, including non-unique values.
      )
```
