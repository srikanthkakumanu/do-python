# Numpy Overview


Numpy is primarily used for numerical operations on arrays. NumPy is essential for data science and scientific computing because it provides tools for handling large datasets efficiently. It is widely used in various fields such as machine learning, data analysis, and engineering.

- **Multi-dimensional arrays**: `ndarray` encapsulates n-dimensional arrays of homogeneous data types. It allows efficient storage and manipulation of large data sets.
- **Performance**: Numpy arrays uses fixed types and are **more efficient than Python lists** because they are stored in contiguous memory locations, which allows for faster access and manipulation. This is known as **locality of reference**.
- **Mathematical Functions**: NumPy provides a wide range of mathematical functions, including operations for *linear algebra, Fourier transforms, and random number generation*.
- **Vectorization and Broadcasting**: NumPy supports vectorized operations, which means that operations are applied element-wise without the need for explicit loops. Broadcasting allows for operations on arrays of different shapes, making the code more concise and readable.


```python
import numpy as np
```

## Basics

```python
# 1D array
a = np.array([1, 2, 3])
print(a)
```

```python
# 2D array
b = np.array([[9.0, 8.0, 7.0], [6.0, 5.0, 4.0]])
print(b)
```

```python
# Get Dimension of an array
print('One dimensional array: ', a.ndim)
print('Two dimensional array: ', b.ndim)
```

```python
# Get Shape of an array
print('Shape of one dimensional array (1 row + 3 columns): ', a.shape)
print('Shape of two dimensional array (2 rows + 3 columns): ', b.shape)
```

```python
# Get data type of an array
c = np.array([1, 2, 3, 4, 5, 6], dtype='int8') # We can specify the data type of an array
print(c.dtype)
print(f'a data type: {a.dtype}, b data type: {b.dtype}, c data type: {c.dtype}') # or print(a.dtype)
```

```python
# Get size of an array
print('a - Size of one dimensional array: ', a.itemsize)
print('b - Size of two dimensional array: ', b.itemsize)
print('c - Size of one dimensional array: ', c.itemsize)
c = np.array([1, 2, 3, 4], dtype='int32')
print("c - Resized size of one dimensional array: ", c.itemsize)
# Total size of an array
print('a - Total size of one dimensional array: ', a.size)
print('b - Total size of two dimensional array: ', b.size)
print('c - Total size of one dimensional array: ', c.size)
# Total size of an array in bytes
print('a - Total size of one dimensional array in bytes (a.itemsize * a.size): ', a.nbytes)
print('b - Total size of two dimensional array in bytes (b.itemsize * b.size): ', b.nbytes)
print('c - Total size of one dimensional array in bytes (c.itemsize * c.size): ', c.nbytes)
```
