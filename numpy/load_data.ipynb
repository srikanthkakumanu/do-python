## Load data from file

```python
import numpy as np
```

```python
data = np.genfromtxt('data.txt', delimiter=',')
print(data)
data = np.genfromtxt('data_with_heading.txt', delimiter=',', skip_header=1)
print(data)
data = np.genfromtxt('data_with_heading.txt', delimiter=',', skip_header=1, usecols=[0, 2])
print(data)
data = np.genfromtxt("data.txt", delimiter=",")
data = data.astype('int32')
print(data)
```
